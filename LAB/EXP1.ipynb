```python
# ==========================================
# FIND-S Algorithm in Google Colab
# ==========================================

import pandas as pd

# Step 1: Create the training dataset
data = {
    'Sky': ['Sunny', 'Sunny', 'Rainy', 'Sunny'],
    'AirTemp': ['Warm', 'Warm', 'Cold', 'Warm'],
    'Humidity': ['Normal', 'High', 'High', 'High'],
    'Wind': ['Strong', 'Strong', 'Strong', 'Strong'],
    'Water': ['Warm', 'Warm', 'Warm', 'Cool'],
    'Forecast': ['Same', 'Same', 'Change', 'Same'],
    'EnjoySport': ['Yes', 'Yes', 'No', 'Yes']
}

df = pd.DataFrame(data)

print("Training Dataset:\n")
print(df)

# Step 2: Separate features and target
concepts = df.iloc[:, :-1].values
target = df.iloc[:, -1].values

# Step 3: FIND-S Algorithm
def find_s(concepts, target):
    # Initialize hypothesis with first positive example
    hypothesis = None

    for i in range(len(target)):
        if target[i] == "Yes":
            if hypothesis is None:
                hypothesis = concepts[i].copy()
            else:
                for j in range(len(hypothesis)):
                    if hypothesis[j] != concepts[i][j]:
                        hypothesis[j] = "?"

            print(f"\nAfter Positive Example {i+1}:")
            print(hypothesis)

    return hypothesis

# Step 4: Train the algorithm
final_hypothesis = find_s(concepts, target)

# Step 5: Display Final Hypothesis
print("\n==============================")
print("Final Most Specific Hypothesis")
print("==============================")
print(final_hypothesis)
```
