```python
# ==========================================
# Candidate Elimination Algorithm
# Google Colab Implementation
# ==========================================

import pandas as pd

# Upload CSV file in Google Colab
from google.colab import files
uploaded = files.upload()

# Read dataset
df = pd.read_csv("training_data.csv")

print("Training Dataset:\n")
print(df)

# Features and Target
concepts = df.iloc[:, :-1].values
target = df.iloc[:, -1].values

# Initialize Specific Hypothesis
S = concepts[0].copy()

# Initialize General Hypothesis
G = [["?" for i in range(len(S))]]

print("\nInitial Specific Hypothesis:")
print(S)

print("\nInitial General Hypothesis:")
print(G)


# Candidate Elimination Algorithm
for i, h in enumerate(concepts):

    if target[i] == "Yes":
        # Positive Example
        for x in range(len(S)):
            if h[x] != S[x]:
                S[x] = "?"
                G[0][x] = "?"

    else:
        # Negative Example
        for x in range(len(S)):
            if h[x] != S[x]:
                G[0][x] = S[x]
            else:
                G[0][x] = "?"

    print("\nStep", i+1)
    print("Specific Hypothesis:", S)
    print("General Hypothesis :", G)

print("\n===================================")
print("Final Specific Hypothesis (S)")
print(S)

print("\nFinal General Hypothesis (G)")
print(G)
```
